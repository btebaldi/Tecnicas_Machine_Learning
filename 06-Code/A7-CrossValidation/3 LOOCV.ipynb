### Validação Cruzada Leave-One-Out (LOOCV)
Este bloco usa trainControl com method="LOOCV". LOOCV é um caso especial de k-fold onde k é igual ao número de observações (n).


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import LeaveOneOut, cross_val_score
from sklearn.datasets import load_iris
from sklearn.linear_model import LinearRegression
```

```python
# Carregar dados (Equivalente a data(iris))
iris_data = load_iris()
iris = pd.DataFrame(data=iris_data.data, columns=iris_data.feature_names)
iris['Species'] = iris_data.target_names[iris_data.target]
```

```python
iris.head()
```

```python
# O modelo de Regressão Linear (lm) requer que a variável de saída (Species) seja numérica.
# No Python, vamos usar a codificação Label Encoding para a variável target.
iris['Species_Numeric'] = pd.factorize(iris['Species'])[0]
X = iris.drop(['Species', 'Species_Numeric'], axis=1)
y = iris['Species_Numeric']

```

```python
iris.describe(include='all')
```

```python

# Validação Cruzada Leave-One-Out (LOOCV)
train_control_loocv = LeaveOneOut()
model_loocv = LinearRegression()
```

```python
scores_loocv = cross_val_score(model_loocv, X, y, cv=train_control_loocv, scoring='neg_mean_squared_error')

print("\n--- Validação Cruzada Leave-One-Out (LOOCV) ---")
print(f"Média do Erro Quadrático Médio Negativo (MSE): {np.mean(scores_loocv):.4f}")
print(f"Número de observações (k=n): {len(scores_loocv)}")
```

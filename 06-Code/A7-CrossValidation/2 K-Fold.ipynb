### Validação Cruzada K-Fold (k=10)
Este bloco usa trainControl com method="cv" (Cross-Validation) e number=10 para avaliar um modelo de Regressão Linear (lm).

Tarefa	R (caret)	Python (sklearn)
Validação Cruzada	trainControl(method="cv")	KFold ou cross_val_score
Modelo de Regressão	train(..., method="lm")	LinearRegression

Como o alvo (Species) é categórico, a Regressão Linear não é a escolha ideal, mas mantemos o modelo linear pela simplicidade educacional. Usaremos o *cross_val_score* para realizar a avalidação cruzada.

```python
# Bibliotecas
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression

from sklearn.datasets import load_iris
# from sklearn.model_selection import train_test_split
# from sklearn.neighbors import KNeighborsClassifier
# from sklearn.metrics import accuracy_score
```

```python
# Carregar dados (Equivalente a data(iris))
iris_data = load_iris()
iris = pd.DataFrame(data=iris_data.data, columns=iris_data.feature_names)
iris['Species'] = iris_data.target_names[iris_data.target]
```

```python
# O modelo de Regressão Linear (lm) requer que a variável de saída (Species) seja numérica.
# No Python, vamos usar a codificação Label Encoding para a variável target.
iris['Species_Numeric'] = pd.factorize(iris['Species'])[0]
X = iris.drop(['Species', 'Species_Numeric'], axis=1)
y = iris['Species_Numeric']

# Validação cruzada (k-fold = 10)
train_control = KFold(n_splits=10, shuffle=True, random_state=42)
model = LinearRegression()
```

```python
# Utilizo o metodo cross_val_score para avaliar o modelo.
scores = cross_val_score(model, X, y, cv=train_control, scoring='neg_mean_squared_error')

print("--- Validação Cruzada K-Fold (k=10) ---")
# O R retorna o resultado do modelo e métricas. Aqui mostramos a média das métricas.
print(f"Média do Erro Quadrático Médio Negativo (MSE): {np.mean(scores):.4f}")
print(f"Desvio Padrão do MSE: {np.std(scores):.4f}")
```

### Validação Cruzada Repetida (Repeated CV)
Este bloco usa trainControl com method="repeatedcv", number=10, e repeats=3.


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import RepeatedKFold, cross_val_score
from sklearn.datasets import load_iris
from sklearn.linear_model import LinearRegression
```

```python
# Carregar dados (Equivalente a data(iris))
iris_data = load_iris()
iris = pd.DataFrame(data=iris_data.data, columns=iris_data.feature_names)
iris['Species'] = iris_data.target_names[iris_data.target]
```

```python
# O modelo de Regressão Linear (lm) requer que a variável de saída (Species) seja numérica.
# No Python, vamos usar a codificação Label Encoding para a variável target.
iris['Species_Numeric'] = pd.factorize(iris['Species'])[0]
X = iris.drop(['Species', 'Species_Numeric'], axis=1)
y = iris['Species_Numeric']
```

```python
# Validação Cruzada Repetida (10-fold, 3 repetições)
train_control_rep = RepeatedKFold(n_splits=10, n_repeats=3, random_state=42)
model_rep = LinearRegression()
```

```python
scores_rep = cross_val_score(model_rep, X, y, cv=train_control_rep, scoring='neg_mean_squared_error')

print("\n--- Validação Cruzada Repetida (10-fold, 3 repetições) ---")

# O número total de avaliações é n_splits * n_repeats = 30
print(f"Total de iterações de CV: {len(scores_rep)}")
print(f"Média do Erro Quadrático Médio Negativo (MSE): {np.mean(scores_rep):.4f}")
```

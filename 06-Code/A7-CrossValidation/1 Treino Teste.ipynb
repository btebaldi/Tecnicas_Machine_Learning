### Divisão de Dados e Classificação KNN
Este bloco usa a função createDataPartition para dividir os dados e treinar um modelo KNN (k-Nearest Neighbors).


```python
#  Lista de bibliotecas utilizadas
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
# Carregar dados (Equivalente a data(iris))
iris_data = load_iris()
iris = pd.DataFrame(data=iris_data.data, columns=iris_data.feature_names)
iris['Species'] = iris_data.target_names[iris_data.target]
```

```python
# Dividir os dados em treino (80%) e teste (20%)
X = iris.drop('Species', axis=1)
y = iris['Species']

# Equivalent to createDataPartition (80/20 split)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=123, stratify=y
)
```

```python
# Treinar o modelo (KNN) - o 'train' do caret com 'knn' usa o k=5 por padrão no scikit-learn
# No caret, o 'train' otimiza k. Aqui usamos um k fixo para simplificar.
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)

# Fazer previsões no conjunto de teste
predictions = model.predict(X_test)
```

```python
print("--- KNN (k=5) ---")
print(f"Previsões no conjunto de teste: {predictions[:5]}...")
print(f"Acurácia: {accuracy_score(y_test, predictions):.4f}")
```

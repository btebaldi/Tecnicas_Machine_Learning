```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
file_path = "ionosphere.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
print("=" * 70)
print("PASSO 1 - Coleta e exploracao dos dados")
print("=" * 70)
print(f"Numero de instancias: {df.shape[0]}")
print(f"Numero de atributos (excluindo a classe): {df.shape[1] - 1}")
print("\nDistribuicao da variavel alvo (y):")
print(df["y"].value_counts().rename({"g": "good", "b": "bad"}))
print()
```

```python
print("=" * 70)
print("PASSO 2 - Preparacao dos dados")
print("=" * 70)

X = df.drop(columns=["y"]).copy()
y = df["y"].map({"g": "good", "b": "bad"})

constant_cols = X.columns[X.nunique() <= 1].tolist()
print(f"Atributo(s) constante(s) detectado(s): {constant_cols}")
X = X.drop(columns=constant_cols)
print(f"Atributos restantes: {X.shape[1]}\n")
```

```python
print("=" * 70)
print("PASSO 3 - Treinamento do modelo (conjunto de treino)")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split( X, y, random_state=42, stratify=y )

model = GaussianNB()
model.fit(X_train, y_train)

priors = pd.Series(model.class_prior_, index=model.classes_)
print("Probabilidades a priori (apriori) estimadas:")
print(priors.round(3))
print()

model, X_train, X_test, y_train, y_test
```

```python
print("=" * 70)
print(f"PASSO 4 - Avaliacao no conjunto de teste")
print("=" * 70)

probs = model.predict_proba(X_test)
print("Primeiras probabilidades previstas (classes: " + f"{list(model.classes_)}):")
print(pd.DataFrame(probs, columns=model.classes_).head(4))

preds = model.predict(X_test)
print(f"Primeiras classes previstas: {preds[:4]}")
```

```python
print("=" * 70)
print(f"PASSO 5 - Metricas de desempenho do modelo")
print("=" * 70)

cm = confusion_matrix(y_test, preds, labels=model.classes_)
cm_df = pd.DataFrame(
    cm, index=[f"real_{c}" for c in model.classes_],
    columns=[f"previsto_{c}" for c in model.classes_]
)
print("\nMatriz de confusao:")
print(cm_df)

acc = accuracy_score(y_test, preds)
print(f"\nAcuracia: {acc:.3f}\n")
```
